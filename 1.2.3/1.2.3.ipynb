```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xs = np.array([0, 4, 16, 36, 64])
ys = np.array([5.39, 5.79, 6.83, 8.70, 11.1])
dy = np.array([0.1]*5)
coeffs = np.polyfit(xs, ys, deg=1)
poly_func = np.poly1d(coeffs)
x_fit = np.linspace(0, max(xs)+5, 200)
y_fit = poly_func(x_fit)
plt.xlabel('Квадрат смещение, см^2')
plt.ylabel('Момент инерции, г*м^2')
plt.grid(linestyle=':')
plt.plot(x_fit, y_fit, label=f"Аппроксимация: {coeffs[0]:.3f}x + {coeffs[1]:.3f}", color="red")
plt.errorbar(xs, ys, yerr=dy, fmt='.', label='Данные')
plt.legend()
plt.title("Зависимость момента инерции от квадрата смещения долек")
plt.show()
```
