```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xs = np.array([245, 491, 734, 982, 1228, 1473, 1719, 1964, 2210, 2455])
ys = np.array([23.6, 25, 26.4, 27.6, 28.7, 30, 31.3, 32.4, 33.6, 34.9])
dy = np.array([0.2]*10)
coeffs = np.polyfit(xs, ys, deg=1)
poly_func = np.poly1d(coeffs)
x_fit = np.linspace(0, max(xs)+100, 200)
y_fit = poly_func(x_fit)
plt.xlabel('Масса, г')
plt.ylabel('показания шкалы, см')
plt.grid(linestyle=':')
plt.plot(x_fit, y_fit, label=f"Аппроксимация: {coeffs[0]:.3f}x + {coeffs[1]:.3f}", color="red")
plt.errorbar(xs, ys, yerr=dy, fmt='.', label='Данные')
plt.legend()
plt.title("Данные полученные с прибора Лермантова")
plt.show()
```
