```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
xs = np.array([39.63, 31.6, 25.39, 16.73])
ys = np.array([20.6, 16.4, 13.1, 8.7])
dy = np.array([0.03, 0.05, 0.04, 0.03])
coeffs = np.polyfit(xs, ys, deg=1)
poly_func = np.poly1d(coeffs)
x_fit = np.linspace(0, max(xs)+5, 200)
y_fit = poly_func(x_fit)
plt.xlabel('Момент силы, Н*см')
plt.ylabel('Частота враения, 100/с')
plt.grid(linestyle=':')
plt.plot(x_fit, y_fit, label=f"Аппроксимация: {coeffs[0]:.3f}x - {abs(coeffs[1]):.3f}", color="red")
plt.errorbar(xs, ys, yerr=dy, fmt='.', label='Данные')
plt.legend()
plt.show()
```
